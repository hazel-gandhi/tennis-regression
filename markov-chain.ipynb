```python
%%R
get_serve_win_prob <- function(row, server = "winner") {
  if (server == "winner") {
    first_in <- row[["w_1stIn"]]
    first_won <- row[["w_1stWon"]]
    second_won <- row[["w_2ndWon"]]
    total_pts <- row[["w_svpt"]]
  } else if (server == "loser") {
    first_in <- row[["l_1stIn"]]
    first_won <- row[["l_1stWon"]]
    second_won <- row[["l_2ndWon"]]
    total_pts <- row[["l_svpt"]]
  } else {
    stop("server must be 'winner' or 'loser'")
  }

  if (is.na(first_in) || is.na(first_won) || is.na(second_won) || is.na(total_pts) ||
      first_in == 0 || total_pts == 0 || (total_pts - first_in) == 0) {
    return(NA)  # Skip invalid or missing data
  }

  first_pct <- first_in / total_pts
  first_win_pct <- first_won / first_in
  second_pts <- total_pts - first_in
  second_win_pct <- second_won / second_pts

  p_win <- (first_pct * first_win_pct) + ((1 - first_pct) * second_win_pct)
  return(p_win)
}
```
