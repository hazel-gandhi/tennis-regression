```python
import pandas as pd 

# show all columns
pd.set_option('display.max_columns', None)

df_raw = pd.read_csv("atp_matches_2023.csv", parse_dates = ['tourney_date'])\
    .sort_values(by='tourney_date')
df_raw
```

```python
from tqdm.auto import tqdm
tqdm.pandas()
```

```python
predictors = []

for index, match in tqdm(df_raw.iterrows(), total=len(df_raw)):
    
    previous_matches = df_raw[df_raw['tourney_date'] < match['tourney_date']]

    # create a new variable
    number_of_previous_matches_won = len(previous_matches.query(f"winner_id == {match['winner_id']}"))

    # append to dataframe
    predictors.append({
        'prev_match_won': number_of_previous_matches_won
    })

pd.DataFrame(predictors)
```

```python
df_predictors = pd.DataFrame(predictors)
df_predictors
```

```python
df = df_raw
```

```python
assert(len(df) == len(df_predictors))
```

```python
pd.concat([df, df_predictors], axis=1)
```
