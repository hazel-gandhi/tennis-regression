```python
import pandas as pd 

# show all columns
pd.set_option('display.max_columns', None)

df_raw = pd.read_csv("atp_matches_2023.csv", parse_dates = ['tourney_date'])\
    .sort_values(by='tourney_date')
df_raw
```

```python
#filtering matches of top 20 players
top_20_players = ['Jannik Sinner', 'Alexander Zverev', 'Carlos Alcaraz', 'Taylor Fritz', 'Novak Djokovic', 'Casper Ruud',
                  'Jack Draper', 'Stefanos Tsitsipas', 'Andrey Rublev', 'Alex De Minaur', 'Daniil Medvedev', 'Holger Rune', 
                  'Tommy Paul', 'Ben Shelton', 'Arthur Fils', 'Lorenzo Musetti', 'Frances Tiafoe', 'Grigor Dimitrov', 
                  'Felix Auger Aliassime', 'Ugo Humbert']
```

```python
#filtering to all matches of top 20 players
df_raw = df_raw[df_raw['winner_name'].isin(top_20_players) | df_raw['loser_name'].isin(top_20_players)]
df_raw
```

```python
# Create two dataframes: one for where top 20 players have won and one where they have lost
wins_df = df_raw[df_raw['winner_name'].isin(top_20_players)].copy()
losses_df = df_raw[df_raw['loser_name'].isin(top_20_players)].copy()

# For wins, I will first make a result column with 1 or 0, in this df all rows in the result column will have 1
wins_df['result'] = 1

#also adding the name of the player who won as the main player for that match
wins_df['player'] = wins_df['winner_name']

#calculating rate of 1st serves won
wins_df['rate_1stWon'] = wins_df['w_1stWon'] / wins_df['w_1stIn']

#calculating rate of break points saved (stat explained in memo)
wins_df['bpSaved_rate'] = wins_df['w_bpSaved'] / wins_df['w_bpFaced']

# for every player, i also want return points stats, and these I will get from the loser's service points stats
wins_df['return_points_won'] = wins_df['l_svpt'] - wins_df['l_1stWon'] - wins_df['l_2ndWon'] #calculating how many return points they faced
wins_df['return_rate'] = wins_df['return_points_won'] / wins_df['l_svpt'] #rate of return points won by winner


#Doing the same calculation for the matches the players lost
losses_df['result'] = 0
losses_df['player'] = losses_df['loser_name']
losses_df['rate_1stWon'] = losses_df['l_1stWon'] / losses_df['l_1stIn']
losses_df['bpSaved_rate'] = losses_df['l_bpSaved'] / losses_df['l_bpFaced']
losses_df['return_points_won'] = losses_df['w_svpt'] - losses_df['w_1stWon'] - losses_df['w_2ndWon']
losses_df['return_rate'] = losses_df['return_points_won'] / losses_df['w_svpt']

#combining the two dataframes so now i have a new dataset of all top 20 players and the matches they won and lost in 2023
combined_df = pd.concat([wins_df, losses_df], ignore_index=True)

combined_df
```

```python
#calcuating null values
null_counts = combined_df[['return_rate', 'bpSaved_rate', 'rate_1stWon']].isnull().sum()
print("Count of null values in each column:")
print(null_counts)
```

#### Adding 0s to null values. 
#### Reason for adding null values explained in memo, though not sure if these are messing up my analysis. they're a very small number though. Maybe they matter when it comes to bpSaved_rate because I converted 115 null values to 0

```python

combined_df['return_rate'] = combined_df['return_rate'].fillna(0)
combined_df['bpSaved_rate'] = combined_df['bpSaved_rate'].fillna(0)
combined_df['rate_1stWon'] = combined_df['rate_1stWon'].fillna(0)
combined_df
```

```python
from tqdm.auto import tqdm
tqdm.pandas()
```

```python
predictors = []

for index, match in tqdm(combined_df.iterrows(), total=len(combined_df)):
    
    player = match['player']
    match_date = match['tourney_date']
    
    # Filter all matches that happened before this one
    previous_matches = combined_df[
        (combined_df['tourney_date'] < match_date) &
        (combined_df['player'] == player)
    ]
    
    # Count previous wins using the 'result' column
    prev_match_won = previous_matches['result'].sum()

    # Average rate_1stWon from previous matches. taking the mean value because count or sum will not be right, like for the previous one 
    avg_rate_1stWon_so_far = previous_matches['rate_1stWon'].dropna().mean()
    avg_rate_1stWon_so_far = avg_rate_1stWon_so_far if not pd.isna(avg_rate_1stWon_so_far) else 0

    #same for bpSaved_rate from previous matches
    avg_bpSaved_rate_so_far = previous_matches['bpSaved_rate'].dropna().mean()
    avg_bpSaved_rate_so_far = avg_bpSaved_rate_so_far if not pd.isna(avg_bpSaved_rate_so_far) else 0

    #same for return_rate
    avg_return_rate_so_far = previous_matches['return_rate'].dropna().mean()
    avg_return_rate_so_far = avg_return_rate_so_far if not pd.isna(avg_return_rate_so_far) else 0

    predictors.append({
        'prev_match_won': prev_match_won,
        'prev_rate_1stWon_avg': avg_rate_1stWon_so_far,
        'prev_bpSaved_rate': avg_bpSaved_rate_so_far,
        'prev_return_rate': avg_return_rate_so_far
    })

predictors_df = pd.DataFrame(predictors)
```
