```python
import pandas as pd 

# show all columns
pd.set_option('display.max_columns', None)

df = pd.read_csv("atp_matches_2023.csv", parse_dates = ['tourney_date'])\
    .sort_values(by='tourney_date')
df
```

```python
# Create two dataframes: one for where top 20 players have won and one where they have lost
wins_df = df.copy()

# For wins, I will first make a result column with 1 or 0, in this df all rows in the result column will have 1
wins_df['result'] = 1

#also adding the name of the player who won as the main player for that match
wins_df['player'] = wins_df['winner_name']

#calculating rate of 1st serves won
wins_df['w_rate_1stWon'] = wins_df['w_1stWon'] / wins_df['w_1stIn'].replace(0, pd.NA)

#calculating rate of break points saved (stat explained in memo)
wins_df['w_bpSaved_rate'] = wins_df['w_bpSaved'] / wins_df['w_bpFaced'].replace(0, pd.NA)

# for every player, i also want return points stats, and these I will get from the loser's service points stats
wins_df['w_return_points_won'] = wins_df['l_svpt'] - wins_df['l_1stWon'] - wins_df['l_2ndWon'] #calculating how many return points they faced
wins_df['w_return_rate'] = wins_df['w_return_points_won'] / wins_df['l_svpt'].replace(0, pd.NA) #rate of return points won by winner


#Doing the same calculation for the matches the players lost
losses_df = df.copy()
losses_df['result'] = 0
losses_df['player'] = losses_df['loser_name']
losses_df['l_rate_1stWon'] = losses_df['l_1stWon'] / losses_df['l_1stIn'].replace(0, pd.NA)
losses_df['l_bpSaved_rate'] = losses_df['l_bpSaved'] / losses_df['l_bpFaced'].replace(0, pd.NA)
losses_df['l_return_points_won'] = losses_df['w_svpt'] - losses_df['w_1stWon'] - losses_df['w_2ndWon']
losses_df['l_return_rate'] = losses_df['l_return_points_won'] / losses_df['w_svpt'].replace(0, pd.NA)

#combining the two dataframes so now i have a new dataset of all top 20 players and the matches they won and lost in 2023
combined_df = pd.concat([wins_df, losses_df], ignore_index=True)

combined_df
```

```python
from tqdm.auto import tqdm
tqdm.pandas()
```

```python
predictors = []

for index, match in tqdm(combined_df.iterrows(), total=len(combined_df)):

    match_date = match['tourney_date']
    
    # Filter all matches that happened before this one
    player = match['player']
    previous_matches = combined_df[
        (combined_df['tourney_date'] < match_date) &
        (combined_df['player'] == player)
        ]

    
    # Count previous wins using the 'result' column
    

    # Average rate_1stWon from previous matches. taking the mean value because count or sum will not be right, like for the previous one 
    w_rate_1stWon_so_far = previous_matches['w_rate_1stWon'].mean()
    l_rate_1stWon_so_far = previous_matches['l_rate_1stWon'].mean()
    # avg_rate_1stWon_so_far = avg_rate_1stWon_so_far if not pd.isna(avg_rate_1stWon_so_far) else 0

    #same for bpSaved_rate from previous matches
    w_bpSaved_rate_so_far = previous_matches['w_bpSaved_rate'].mean()
    l_bpSaved_rate_so_far = previous_matches['l_bpSaved_rate'].mean()
    # avg_bpSaved_rate_so_far = avg_bpSaved_rate_so_far if not pd.isna(avg_bpSaved_rate_so_far) else 0

    #same for return_rate
    w_return_rate_so_far = previous_matches['w_return_rate'].mean()
    l_return_rate_so_far = previous_matches['l_return_rate'].mean()
    # avg_return_rate_so_far = avg_return_rate_so_far if not pd.isna(avg_return_rate_so_far) else 0

    predictors.append({
       'w_prev_rate_1stWon_avg': w_rate_1stWon_so_far,
        'l_prev_rate_1stWon_avg': l_rate_1stWon_so_far,
        'w_prev_bpSaved_rate': w_bpSaved_rate_so_far,
        'l_prev_bpSaved_rate': l_bpSaved_rate_so_far,
        'w_prev_return_rate': w_return_rate_so_far,
        'l_prev_return_rate': l_return_rate_so_far,   
    })

predictors_df = pd.DataFrame(predictors)
predictors_df
```

```python
new_df = pd.concat([combined_df.reset_index(drop=True), predictors_df.reset_index(drop=True)], axis=1)
```
