```python
import pandas as pd
import numpy as np

def transform_tennis_data(input_file, output_file=None):
    """
    Transform tennis match data from winner/loser format to player A/B format.
    Using random assignment for player A and B.
    
    Parameters:
    input_file (str): Path to the input CSV file
    output_file (str, optional): Path to the output CSV file. If None, no file is saved.
    """
    # Read the CSV file
    df = pd.read_csv("atp_matches_2023.csv")
    
    # Use random assignment for player A and B
    np.random.seed(42)  # For reproducibility
    winner_is_A = np.random.randint(0, 2, size=len(df)) == 1
    
    # Create a new dataframe for the transformed data
    result_df = pd.DataFrame()
    
    # Copy non-player columns directly
    non_player_cols = [col for col in df.columns if not (col.startswith('winner_') or 
                                                       col.startswith('loser_') or 
                                                       col.startswith('w_') or 
                                                       col.startswith('l_'))]
    for col in non_player_cols:
        result_df[col] = df[col]
    
    # Map player attributes
    player_attributes = [
        ('id', 'winner_id', 'loser_id'),
        ('seed', 'winner_seed', 'loser_seed'),
        ('entry', 'winner_entry', 'loser_entry'),
        ('name', 'winner_name', 'loser_name'),
        ('hand', 'winner_hand', 'loser_hand'),
        ('ht', 'winner_ht', 'loser_ht'),
        ('ioc', 'winner_ioc', 'loser_ioc'),
        ('age', 'winner_age', 'loser_age'),
        ('rank', 'winner_rank', 'loser_rank'),
        ('rank_points', 'winner_rank_points', 'loser_rank_points')
    ]
    
    for attr, winner_col, loser_col in player_attributes:
        result_df[f'player_A_{attr}'] = np.where(winner_is_A, df[winner_col], df[loser_col])
        result_df[f'player_B_{attr}'] = np.where(winner_is_A, df[loser_col], df[winner_col])
    
    # Map player statistics
    player_stats = [
        ('ace', 'w_ace', 'l_ace'),
        ('df', 'w_df', 'l_df'),
        ('svpt', 'w_svpt', 'l_svpt'),
        ('1stIn', 'w_1stIn', 'l_1stIn'),
        ('1stWon', 'w_1stWon', 'l_1stWon'),
        ('2ndWon', 'w_2ndWon', 'l_2ndWon'),
        ('SvGms', 'w_SvGms', 'l_SvGms'),
        ('bpSaved', 'w_bpSaved', 'l_bpSaved'),
        ('bpFaced', 'w_bpFaced', 'l_bpFaced')
    ]
    
    for stat, winner_col, loser_col in player_stats:
        result_df[f'player_A_{stat}'] = np.where(winner_is_A, df[winner_col], df[loser_col])
        result_df[f'player_B_{stat}'] = np.where(winner_is_A, df[loser_col], df[winner_col])
    
    # Add winner column 
    result_df['winner'] = np.where(winner_is_A, 'A', 'B')
    
    # Add player_A_win column (0 or 1)
    result_df['player_A_win'] = winner_is_A.astype(int)
    
    # Save the transformed data if output_file is provided
    if output_file:
        result_df.to_csv(output_file, index=False)
    
    return result_df

# Example usage with random assignment
df_transformed = transform_tennis_data('atp_matches_2023.csv', 'atp_matches_2023_transformed.csv')
df_transformed
df = df_transformed
df
```

```python

#calculating rate of 1st serves won
df['player_A_rate_1stWon'] = df['player_A_1stWon'] / df['player_A_1stIn'].replace(0, pd.NA)

#calculating rate of break points saved (stat explained in memo)
df['player_A_bpSaved_rate'] = df['player_A_bpSaved'] / df['player_A_bpFaced'].replace(0, pd.NA)

# for every player, i also want return points stats, and these I will get from the loser's service points stats
df['player_A_return_points_won'] = df['player_B_svpt'] - df['player_B_1stWon'] - df['player_B_2ndWon'] #calculating how many return points they faced
df['player_A_return_rate'] = df['player_A_return_points_won'] / df['player_B_svpt'].replace(0, pd.NA) #rate of return points won by winner


#Doing the same calculation for the matches the players lost
df['player_B_rate_1stWon'] = df['player_B_1stWon'] / df['player_B_1stIn'].replace(0, pd.NA)
df['player_B_bpSaved_rate'] = df['player_B_bpSaved'] / df['player_B_bpFaced'].replace(0, pd.NA)
df['player_B_return_points_won'] = df['player_A_svpt'] - df['player_A_1stWon'] - df['player_A_2ndWon']
df['player_B_return_rate'] = df['player_B_return_points_won'] / df['player_B_svpt'].replace(0, pd.NA)

#combining the two dataframes so now i have a new dataset of all top 20 players and the matches they won and lost in 2023
df
```

```python
predictors = []

for index, match in tqdm(df.iterrows(), total=len(df)):

    match_date = match['tourney_date']
    
    # Filter all matches that happened before this one
    # player = match['player']
    previous_matches = df[
        (df['tourney_date'] < match_date)
        # (combined_df['player'] == player)
        ]

    
    # Count previous wins using the 'result' column
    

    # Average rate_1stWon from previous matches. taking the mean value because count or sum will not be right, like for the previous one 
    player_A_rate_1stWon_so_far = previous_matches['player_A_rate_1stWon'].mean(skipna=True) or 0
    player_B_rate_1stWon_so_far = previous_matches['player_B_rate_1stWon'].mean(skipna=True) or 0
    # avg_rate_1stWon_so_far = avg_rate_1stWon_so_far if not pd.isna(avg_rate_1stWon_so_far) else 0

    #same for bpSaved_rate from previous matches
    player_A_bpSaved_rate_so_far = previous_matches['player_A_bpSaved_rate'].mean(skipna=True) or 0
    player_B_bpSaved_rate_so_far = previous_matches['player_B_bpSaved_rate'].mean(skipna=True) or 0
    # avg_bpSaved_rate_so_far = avg_bpSaved_rate_so_far if not pd.isna(avg_bpSaved_rate_so_far) else 0

    #same for return_rate
    player_A_return_rate_so_far = previous_matches['player_A_return_rate'].mean(skipna=True) or 0
    player_B_return_rate_so_far = previous_matches['player_B_return_rate'].mean(skipna=True) or 0
    # avg_return_rate_so_far = avg_return_rate_so_far if not pd.isna(avg_return_rate_so_far) else 0

    predictors.append({
       'player_A_prev_rate_1stWon_avg': player_A_rate_1stWon_so_far,
        'player_B_prev_rate_1stWon_avg': player_B_rate_1stWon_so_far,
        'player_A_prev_bpSaved_rate': player_A_bpSaved_rate_so_far,
        'player_B_prev_bpSaved_rate': player_B_bpSaved_rate_so_far,
        'player_A_prev_return_rate': player_A_return_rate_so_far,
        'player_B_prev_return_rate': player_B_return_rate_so_far,   
    })

predictors_df = pd.DataFrame(predictors)
predictors_df
```
